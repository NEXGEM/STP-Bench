# 4. Visualizing Predictions

`stp.visualize(gene, sample, ...)` renders one gene's predicted expression
for one sample as a heatmap and saves it to a PNG.


> **Prerequisites**
> - STP-Bench installed (`bash scripts/create_env.sh`, see the
>   [README](../README.md#installation)).
> - Benchmark data downloaded for the dataset(s) used below (see
>   [README — Benchmark Data](../README.md#benchmark-data)), or your own
>   dataset added following
>   [docs/guide.md — Adding a New Dataset](../docs/guide.md#adding-a-new-dataset).
> - Run this notebook from the repo root, or pass `repo_root=` explicitly to
>   `STPred(...)`.


```python
from stpbench import STPred

stp = STPred(models=["StNet"], repo_root=".")
stp.predict(data="/path/to/slide.svs", output_dir="/path/to/output", train_data="ncche/xenium")
```

```python
path = stp.visualize(gene="SFTPB", sample="TENX118")
path
```

`output_dir` defaults to the most recent `predict()` call's output directory
on this instance, so it can be omitted right after a `predict()` call like
above.

If the original WSI can still be located (via the `predict()` run's own
config — which is automatic right after a WSI-path `predict()` call), the
heatmap is overlaid on the slide's own thumbnail with each patch drawn at
its true non-overlapping footprint. Otherwise `visualize()` falls back to a
plain spatial scatter plot and prints a note to stderr rather than failing.


## Disambiguating between models

If more than one model predicted the same sample into the same
`output_dir`, pass `model=` to pick which one's prediction to visualize —
otherwise `visualize()` raises, listing the candidates.


```python
stp.visualize(gene="SFTPB", sample="TENX118", model="TRIPLEX")
```

## Explicit `output_dir` and `save_path`

Useful when visualizing a prediction from an earlier session rather than
one just produced in this notebook.


```python
stp.visualize(
    gene="SFTPB",
    sample="TENX118",
    output_dir="/path/to/output",
    save_path="/path/to/figures/TENX118_SFTPB.png",
)
```
